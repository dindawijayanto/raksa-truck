# Model B — Physics-Informed Synthetic ML Benchmark

Notebook ini membangun benchmark simulasi truk, melatih surrogate ML untuk RUL dan service horizon, lalu menyimpan weight di `models/model_b/synthetic_ml_v1`. Semua metrik hanya valid terhadap label simulasi.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / 'src').exists():
    PROJECT_ROOT = next(parent for parent in PROJECT_ROOT.parents if (parent / 'src').exists())
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

from aic.synthetic_ml import (
    SyntheticBenchmarkConfig, generate_synthetic_benchmark, train_synthetic_ml,
    write_synthetic_benchmark, write_synthetic_ml_outputs,
)

config = SyntheticBenchmarkConfig()
benchmark = generate_synthetic_benchmark(config)
write_synthetic_benchmark(benchmark, PROJECT_ROOT / 'data' / 'synthetic' / 'model_b_physics_benchmark.csv', config)
benchmark.head()
```

```python
result = train_synthetic_ml(benchmark, config)
write_synthetic_ml_outputs(
    result, benchmark,
    PROJECT_ROOT / 'models' / 'model_b' / 'synthetic_ml_v1',
    PROJECT_ROOT / 'reports' / 'model_b_synthetic_ml',
)
result.metrics
```
